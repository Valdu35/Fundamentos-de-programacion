Ejercicio 1 Área de un triángulo

```python
base = float(input("Base del triangulo: "))
altura = float(input("Altura del triangulo: "))
area = (base * altura) / 2
print("El área del triángulo es:", area)
```

Ejercicio 2 Conversión de Celsius a Fahrenheit

```python
celsius = float(input("Grados Celsius: "))
fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius:g} grados Celsius equivalen a {fahrenheit} grados Fahrenheit")
```

Ejercicio 5 Casting tras 'input()'

```python
x = input("Ingresa un número: ")
print("x es tipo", type(x).__name__)
try:
    x + 1
except TypeError:
    print("x + 1 genera TypeError")
print("int(x) + 1 =", int(x) + 1)
```

Ejercicio 7 Descuento del 15% con f-strings

Ejercicio 9 Mayor de tres numeros

```python
a = int(input("Primer número: "))
b = int(input("Segundo número: "))
c = int(input("Tercer número: "))
if a >= b and a >= c:
    mayor = a
elif b >= c:
    mayor = b
else:
    mayor = c
print("El mayor de los tres números es:", mayor)
```

Ejercicio 10 Suma acumulada con 'while'

```python
suma = 0
numero = int(input("Número (0 para terminar): "))
while numero != 0:
    suma += numero
    numero = int(input("Número (0 para terminar): "))
print("La suma de los números es:", suma)
```

Ejercicio 12 Suma de pares del 1 al 20 con 'for' y 'continue'

```python
suma = 0
for i in range(1, 21):
    if i % 2 != 0:
        continue
    suma += i
print("La suma de los números pares del 1 al 20 es:", suma)
```

Ejercicio 13  Suma de una lista con funcion que regresa valor 

```python
def suma_lista(lista):
    total = 0
    for elemento in lista:
        total += elemento
    return total

print("La suma de los elementos es:", suma_lista([10, 20, 30, 40]))
```

Ejercicio 14 Maximo de una lista con funcion que regresa valor 

```python
def maximo(lista):
    mayor = lista[0]
    for elemento in lista:
        if elemento > mayor:
            mayor = elemento
    return mayor

print("El elemento máximo es:", maximo([7, 3, 9, 2]))
```

Ejercicio 15 Matriz y suma de la diagonal principal

```python
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
suma = 0
for i in range(3):
    suma += matriz[i][i]
print("La suma de la diagonal principal es:", suma)
```

Ejercicio 18 Factorial con funcion recursiva 

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print("El factorial de 5 es:", factorial(5))
```

Ejercicio 19 Tupla: Acceso y suma con funcion 

```python
def suma_tupla(tupla):
    total = 0
    for elemento in tupla:
        total += elemento
    return total

numeros = (4, 8, 15, 16, 23, 42)
print("El tercer elemento es:", numeros[2])
print("La suma de la tupla es:", suma_tupla(numeros))
```

Ejercicio 21 Diccionario de calificaciones y promedios 

```python
calificaciones = {"Ana": [8, 9, 10], "Beto": [7, 7, 8]}
for alumno, notas in calificaciones.items():
    print(f"{alumno}: {sum(notas) / len(notas):.2f}")
```
