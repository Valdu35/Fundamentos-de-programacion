```python
def sumar_tupla(tupla):
    """Recibe una tupla de números y retorna la suma de todos."""
    return sum(tupla)


def opcion_tuplas():
    print("\n--- SECCIÓN: TUPLAS ---")

    numeros = (10, 20, 30, 40, 50)

    print("Tercer elemento:", numeros[2])

    num1 = float(input("Ingresa un número nuevo 1: "))
    num2 = float(input("Ingresa un número nuevo 2: "))

    nueva_tupla = numeros + (num1, num2)
    print("Nueva tupla:", nueva_tupla)

    lista_numeros = list(nueva_tupla)
    lista_numeros.sort()
    print("Lista ordenada:", lista_numeros)

    print("Suma total:", sumar_tupla(nueva_tupla))
```

```python
def buscar_telefono(diccionario, nombre):
    """Recibe el diccionario y un nombre, retorna el teléfono."""
    return diccionario.get(nombre)


def opcion_diccionarios(contactos):
    print("\n--- SECCIÓN: DICCIONARIOS ---")

    nombre_nuevo = input("Nombre del nuevo contacto: ")
    telefono_nuevo = input("Teléfono del nuevo contacto: ")
    contactos[nombre_nuevo] = telefono_nuevo

    print("Nombres registrados:")
    for nombre in contactos.keys():
        print("-", nombre)

    nombre_buscar = input("¿Qué contacto quieres buscar? ")
    telefono = buscar_telefono(contactos, nombre_buscar)

    if telefono:
        print(f"El teléfono de {nombre_buscar} es: {telefono}")
    else:
        print("Ese contacto no existe en la agenda.")
```

```python
def opcion_excepciones():
    print("\n--- SECCIÓN: EXCEPCIONES ---")

    try:
        num1 = int(input("Ingresa el primer número entero: "))
        num2 = int(input("Ingresa el segundo número entero: "))
        resultado = num1 + num2
        print(f"La suma de {num1} y {num2} es: {resultado}")
    except ValueError:
        print("Error: Debes ingresar únicamente números enteros válidos (no letras ni dejar vacío).")
    except ZeroDivisionError:
        print("Error: No es posible dividir entre cero.")
```

```python
def contar_palabras(texto):
    """Recibe un string y retorna cuántas palabras contiene."""
    return len(texto.split())


def opcion_strings():
    print("\n--- SECCIÓN: STRINGS ---")

    mensaje = input("Escribe un mensaje: ")

    print("Longitud del mensaje:", len(mensaje))
    print("En mayúsculas:", mensaje.upper())

    palabra_buscar = input("¿Qué palabra quieres reemplazar? ")
    palabra_nueva = input("¿Por cuál palabra la quieres reemplazar? ")
    mensaje_reemplazado = mensaje.replace(palabra_buscar, palabra_nueva)
    print("Mensaje con reemplazo:", mensaje_reemplazado)

    print("Cantidad de palabras:", contar_palabras(mensaje))
```

```python
def menu_principal():
    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mía": "555-0103"
    }

    while True:
        print("\n===== MENÚ PRINCIPAL =====")
        print("1. Tuplas")
        print("2. Diccionarios")
        print("3. Excepciones")
        print("4. Strings")
        print("5. Salir")

        opcion = input("Elige una opción (1-5): ")

        if opcion == "1":
            opcion_tuplas()
        elif opcion == "2":
            opcion_diccionarios(contactos)
        elif opcion == "3":
            opcion_excepciones()
        elif opcion == "4":
            opcion_strings()
        elif opcion == "5":
            print("¡Hasta luego!")
            break
        else:
            print("Opción no válida. Intenta de nuevo.")

            if __name__ == "__main__":
    menu_principal()
```
