```python
candidatos = {
    "A": 0,
    "B": 0,
    "C": 0,

}

cantidad = int(input("¿Cuántos votos van a ser?"))

for c in range(1, cantidad +1):
    voto= input(f"Ingresa el voto {c} (A, B, C): ")
    print(f"Votaste por {voto.upper()}")
    if voto.upper() in candidatos:
       print(f"Voto para {voto.upper()}")
       candidatos[voto.upper()] += 1
    else: 
         print("EL voto no es valido.")


for k, v in candidatos.items():
    print(f"Opcion {k}, votos: {v}")
```

```python
candidatos = {
    "A": 0,
    "B": 0,
    "C": 0,

}

cantidad = int(input("¿Cuántos votos van a ser?"))
conteo = 1


while conteo <= cantidad:
    voto= input(f"Ingresa el voto {conteo} (A, B, C): ")
    if voto.upper() in candidatos:
       print(f"Voto valido para {voto.upper()}")
       candidatos[voto.upper()] += 1
       conteo += 1
    else: 
         print("EL voto no es valido, vota nuevamente.")

print("\nResultados:")
for k, v in candidatos.items():
    print(f"Opcion {k}, votos: {v}")
```
