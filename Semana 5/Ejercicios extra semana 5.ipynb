Extra 1

```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

print("Tercera calificación:", calificaciones[2])

nueva1 = float(input("Nueva calificación 1: "))
nueva2 = float(input("Nueva calificación 2: "))

nueva_tupla = calificaciones + (nueva1, nueva2)
print("Nueva tupla:", nueva_tupla)

lista_calificaciones = list(nueva_tupla)
lista_calificaciones.sort(reverse=True) 
print("Lista ordenada (mayor a menor):", lista_calificaciones)

def sumar_lista(lista):
    return sum(lista)

print("Suma total:", sumar_lista(lista_calificaciones))
```

Extra 2

```python
agenda = {
        "Jose": "442-173-3912",
        "Yarah": "442-821-4452",
        "Cristobal": "442-576-4258"
}

nombre_nuevo = input("Nombre del nuevo contacto: ")
telefono_nuevo = input("Teléfono del nuevo contacto: ")
agenda[nombre_nuevo] = telefono_nuevo


print("Contactos registrados:", ", ".join(agenda.keys()))

def buscar_telefono(diccionario, nombre):
    return diccionario.get(nombre)

nombre_buscar = input("Nombre a buscar: ")
telefono_encontrado = buscar_telefono(agenda, nombre_buscar)
print(f"El teléfono de {nombre_buscar} es: {telefono_encontrado}")
```

Ejercicio 3

```python
try:
    num1 = int(input("Primer número entero: "))
    num2 = int(input("Segundo número entero: "))
    resultado = num1 / num2
    print(f"{num1} dividido entre {num2} es: {resultado}")
except ValueError:
    print("Error: Debes ingresar únicamente números enteros válidos.")
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
```

```python
mensaje = "Python es un lenguaje poderoso"

print("Longitud del mensaje:", len(mensaje))
print("En mayúsculas:", mensaje.upper())

texto_reemplazado = mensaje.replace("Python", "programación")
print("Texto reemplazado:", texto_reemplazado)


def contar_palabras(texto):
    return len(texto.split())

print("Palabras totales:", contar_palabras(mensaje))
```
