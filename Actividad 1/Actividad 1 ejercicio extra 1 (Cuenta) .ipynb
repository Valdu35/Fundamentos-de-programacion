Ejercicio extra 1 - Cuenta

```python
cuenta = float(input("Total de la cuenta: "))
propina = float(input("Porcentaje de propina: "))
personas = int(input("Número de personas: "))


monto_propina = cuenta * propina / 100

total = cuenta + monto_propina

pago_persona = total / personas


print(f"Propina: ${monto_propina:.2f}")
print(f"Total con propina: ${total:.2f}")
print(f"Pago por persona: ${pago_persona:.2f}")
```
