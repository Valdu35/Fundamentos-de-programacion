Calculadora horas diarias 

```python
nombre = input("Introduce tu nombre ")

Netflix = float(input("Introduce el numero de horas diarias en Netflix: "))
Instagram = float(input("Introduce el numero de horas diarias en Instagram: "))
TikTok = float(input("Introduce el numero de horas diarias en TikTok: "))
Whatsapp = float(input("Introduce el numero de horas diarias en Whatsapp: "))
CandyCrush = float(input("Introduce el numero de horas diarias en Candy Crush: "))

resultado = (Netflix + Instagram + TikTok + Whatsapp + CandyCrush )

porcentaje = resultado / 24 * 100

print(f"{nombre}, tu tiempo total acumulado es de: {resultado} horas al día")
print(f"Porcentaje de uso diario calculado: {porcentaje}%")
```
