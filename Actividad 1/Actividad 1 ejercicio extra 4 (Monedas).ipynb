Ejercicio extra 4 - Monedas

```python
cantidad = float(input("Cantidad en MXN: "))
dolar = float(input("Tipo de cambio USD: "))
euro = float(input("Tipo de cambio EUR: "))


resultado_dolar = cantidad / dolar
resultado_euro = cantidad / euro


print(f"${cantidad:.2f} MXN equivalen a:")
print(f"USD: {resultado_dolar:.2f}")
print(f"EUR: {resultado_euro:.2f}")
```
