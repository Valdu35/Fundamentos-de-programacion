Ejercicio Extra 3 - Calificaciones

```python
parcial1 = float(input("Parcial 1 (30%): "))
parcial2 = float(input("Parcial 2 (30%): "))
parcial3 = float(input("Parcial 3 (40%): "))


resultado = (parcial1 * 0.30) + (parcial2 * 0.30) + (parcial3 * 0.40)


print(f"Tu calificación final es: {resultado:.2f}")
```
