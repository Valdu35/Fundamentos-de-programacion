Ejercicio extra 2 - Minutos, días, horas

```python
minutos = int(input("Total de minutos: "))


dias = minutos // 1440

minutos_restantes = minutos % 1440

horas = minutos_restantes // 60

minutos_finales = minutos_restantes % 60


print(f"{minutos} minutos = {dias} día(s), {horas} hora(s), {minutos_finales} minuto(s)")
```
