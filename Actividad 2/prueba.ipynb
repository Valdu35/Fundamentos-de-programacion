```python
tabla = int(input("¿Que tabla quieres que muestre"))

# for(Int i; i < 10; i++)
print("\nFor")
for m in range(1,11):
   print(tabla, "x", m, "=", tabla * m)

print("\nwhile")
m = 1
while m <= 10: 
    print(f"{tabla} x {m} = {tabla * m}" )
```
