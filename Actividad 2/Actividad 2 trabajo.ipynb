Cobro entradas museo 

```python


PRECIO_NINO = 0
PRECIO_MENOR = 30
PRECIO_ADULTO = 45

total_visitantes = int(input("Cuantos visitantes son: "))
contador = 0
total_general = 0

while True:
    if contador >= total_visitantes:
        break

    edad = int(input("Edad del visitante: "))

    if edad < 0:
        print("Edad invalida")
        continue

    if edad < 3:
        precio = PRECIO_NINO
    elif edad <= 17:
        precio = PRECIO_MENOR
    else:
        precio = PRECIO_ADULTO

    tipo = "general"
    if edad >= 18:
        tipo = input("Tipo (adulto_mayor / profesor / estudiante / general): ")

    if tipo == "adulto_mayor":
        descuento = precio * 0.12
    elif tipo == "profesor" or tipo == "estudiante":
        descuento = precio * 0.10
    else:
        descuento = 0

    total_individual = precio - descuento
    print(f"Subtotal: ${precio:.2f}  Descuento: ${descuento:.2f}  Total: ${total_individual:.2f}")

    total_general += total_individual
    contador += 1

for i in range(1, contador + 1):
    print(f"Visitante {i} listo")

print(f"Total general a pagar: ${total_general:.2f}")
```
