Trabajo extra 1
Control de aforo 

```python
total_acumulado = 0   
visitante = 0        

while True:
    visitante = visitante + 1 

   
    edad = int(input(f"Edad del visitante {visitante}: "))

   
    if edad < 3:
        costo = 0
    elif edad <= 17:
        costo = 30
    else:
        costo = 45

    print(f"Costo del boleto ${visitante}: {costo}")

    
    if costo == 0:
        print("Menor de 3 años, sin cargo")
        continue

   
    total_acumulado = total_acumulado + costo

    
    if total_acumulado >= 100:
        print("Aforo alcanzado, se detiene el registro")
        print(f"Total acumulado: ${total_acumulado:.2f}")
        break
```

Trabajo extra 2 
Estadisticas de visitantes

```python
num_visitantes = int(input("Número de visitantes: "))

adultos = 0      
suma_edades = 0     

for visitante in range(1, num_visitantes + 1):
    edad = int(input(f"Edad del visitante {visitante}: "))

    
    suma_edades = suma_edades + edad

    
    if edad >= 18:
        adultos = adultos + 1


promedio = suma_edades / num_visitantes

print(f"Adultos: {adultos}")
print(f"Promedio de edad: {promedio:.2f}")
```

Trabajo extra 3 
Depuración de cobro

código original con error:

```python
precio = 45
descuento = 12   
total = precio - descuento
print(f"Total: ${total:.2f}") 
```

código corregido: 

```python
precio = 45
descuento = 0.12 

total = precio - (precio * descuento)  

print("Error de lógica: descuento debe ser el porcentaje 0.12, no 12.")
print(f"Código corregido: descuento = precio * 0.12 -> Total: ${total:.2f}")
```

Trabajo extra 4
Pirámide de asteriscos 

```python
altura = int(input("Altura de la pirámide: "))


for fila in range(1, altura + 1):

    
    for columna in range(fila):
        print("*", end="")  

    print()  
```
